```python
from pyspark.sql.functions import when, col

gold_df = spark.read.table("workspace.stock_data.company_profiles_silver")

gold_df = gold_df.withColumn(
    "market_cap_category",
    when(col("market_cap") >= 10e9, "Large Cap")
    .when((col("market_cap") >= 2e9) & (col("market_cap") < 10e9), "Mid Cap")
    .otherwise("Small Cap")
)

gold_df.write.format("delta").mode("overwrite").saveAsTable('workspace.stock_data.company_profiles_gold')

```
